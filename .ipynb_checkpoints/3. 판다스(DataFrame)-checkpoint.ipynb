# DataFrame
- Series가 1차원이었다면 DataFrame은 2차원으로 확대된 버전이다.
- Excel의 SpreadSheet라고 생각하면 이해하기 쉽다.
- 2차원이기 때문에 인덱스가 row, column으로 구성되어 있다.<br/>
        - row: 각 개별 데이터<br/>
        - column: 각 개별 속성을 의미<br/>
- Data Analtsis, Machine Learning에서 Data를 변경하기 위해서 많이 사용된다.

## 컬럼의 의미

    - class: Passenger Class(1 = 1st, 2 = 2nd, 3 = 3rd)
    - name: Name #이름
    - sex: Sex #성별
    - age: Age #나이
    - sibsp: Number of Siblings/Spouses Aboard #동승 형제
    - parch: Number of Parents/Children Aboard #동승 부모,자식
    - ticket: Ticket Number #티켓 번호
    - fare: Passenger Fare
    - cabin: Cabin # 객실번호

## 여성의 생존자 조회

```python
female = titanic_df["Sex"] == "female"
woman_survived_mask = (titanic_df["Survived"] == 1) & (titanic_df["Sex"] == "female")
titanic_df[woman_survived_mask]
```

```python
print(f"생존자 수: {len(titanic_df[woman_survived_mask])}명")
print(f"생존율: 약 {round(len(titanic_df[woman_survived_mask]) / len(titanic_df[female]), 4) * 100}%")

len(female)
```

## 1등석과 3등석의 생존율 비교

```python
all_1st = titanic_df["Pclass"] == 1
all_3st = titanic_df["Pclass"] == 3
survived_1st = (titanic_df["Survived"] == 1) & (titanic_df["Pclass"] == 1)
survived_3rd = (titanic_df["Survived"] == 1) & (titanic_df["Pclass"] == 3)
print(f"1등석 생존자 수: {len(titanic_df[survived_1st])}")
print(f"3등석 생존자 수: {len(titanic_df[survived_3rd])}")

print(f"1등석 생존율: 약 {round(len(titanic_df[survived_1st]) / len(titanic_df[all_1st]), 3) * 100}%")
print(f"3등석 생존율: 약 {round(len(titanic_df[survived_3rd]) / len(titanic_df[all_3st]), 3) * 100}%")
```

```python
titanic_df.groupby("Pclass")["Survived"].mean()

# 클래스별로 묶어서 각자 한번에 추출
```

```python
titanic_df.groupby("Pclass")["Survived"].agg(["mean","count"])
# 또 다른 방법
```

```python
survived_pclass = titanic_df.groupby("Pclass")["Survived"].mean()
pd.DataFrame(data=survived_pclass * 100, index=[1, 2, 3])

#다른 방법
```

```python
survived_pclass = titanic_df.groupby("Pclass")["Survived"].mean()
survived_pclass.index = ["1등석", "2등석", "3등석"]
survived_pclass

pd.DataFrame(data=survived_pclass * 100, index=["1등석", "2등석", "3등석"])
```

# 파생 변수 
- 기존에 존재하는 속성(변수, 칼럽)으로 부터 새로운 속성(컬럼)을 만들어낸 변수

```python
# key가 없으면 추가
titanic_df["RealAge"] = titanic_df["Age"] - 1
titanic_df
```

# 변수 삭제(column 삭제): .drop()
- axis=0(행), 1(열) - 0 이 default

```python
titanic_df.drop(["Cabin", "Embarked"], axis=1)
```

```python
titanic_df.drop(["Cabin", "Embarked"], axis=1, inplace=True) #원본지우기
```

```python
titanic_df
```

## 파생 변수 만들기 연습

```python
# 만약 NaN이라면: 알 수 없음
# 나이가 19세 이상이면 : 성인
# 19세 미만이면 : 아이

def get_adult(age):
    if np.isnan(age):
        return "알 수 없음"
    return "성인" if age >= 19 else "아이"

get_adult(20)
```

```python
# .apply(): 앞의 시리즈의 값을 순차적으로 받아서 결과를 바꾸고 새로운 결과를 반환하는 함수
titanic_df["IsAdult"] = titanic_df["Age"].apply(get_adult)
titanic_df
```

```python
# Fare > 300 : 바가지요금 이하는 정상. : RipOff
```

```python
titanic_df["RipOff"] = titanic_df["Fare"].apply(lambda fare: "바가지" if fare > 300 else "정상")
titanic_df["RipOff"]
```

```python
titanic_df[titanic_df["Fare"] > 300]
```
