# 케글 HR 이직 데이터 셋
- IBM HR Analytics Employee Attrition & Performance 데이터셋은 기업 내 직원들의 인적 사항, 직무 만족도, 근무 환경 및 수당 등의 정보를 바탕으로 직원의 이직 여부(Attrition)를 분석하고 예측하기 위해 수집된 데이터입니다. 인사 관리자(HR)가 직원들의 이탈 원인을 파악하고 조직 만족도와 인재 유지율을 높이기 위한 비즈니스 인사이트를 도출하는 EDA 및 머신러닝 과제에 자주 활용됩니다.
  
- [IBM HR Analytics Employee Attrition & Performance](https://www.kaggle.com/datasets/pavansubhasht/ibm-hr-analytics-attrition-dataset)

## 데이터셋 표
| 컬럼명 | 설명 | 데이터 형태 | 데이터 타입 |
| :--- | :--- | :--- | :--- |
| Attrition(타겟) | 이직 여부 (Yes/No) | 범주형 (Categorical) | object / string |
| Age | 직원 연령 | 수치형 (Numerical) | int |
| Gender | 성별 (Male/Female) | 범주형 (Categorical) | object / string |
| MaritalStatus | 결혼 여부 (Single, Married, Divorced) | 범주형 (Categorical) | object / string |
| Education | 학력 수준 (1: 이하 ~ 5: 박사) | 범주형 (Categorical) | int |
| Department | 근무 부서 | 범주형 (Categorical) | object / string |
| JobRole | 세부 직무 | 범주형 (Categorical) | object / string |
| BusinessTravel | 출장 빈도 | 범주형 (Categorical) | object / string |
| OverTime | 야근 여부 (Yes/No) | 범주형 (Categorical) | object / string |
| DistanceFromHome | 출퇴근 거리 | 수치형 (Numerical) | int |
| MonthlyIncome | 월소득 | 수치형 (Numerical) | int |
| PercentSalaryHike | 전년 대비 인상률 (%) | 수치형 (Numerical) | int |
| StockOptionLevel | 스톡옵션 등급 (0~3) | 범주형 (Categorical) | int |
| TotalWorkingYears | 총 경력 연수 | 수치형 (Numerical) | int |
| YearsAtCompany | 현 회사 근속 연수 | 수치형 (Numerical) | int |
| EnvironmentSatisfaction | 근무 환경 만족도 (1~4) | 범주형 (Categorical) | int |
| JobSatisfaction | 직무 만족도 (1~4) | 범주형 (Categorical) | int |
| WorkLifeBalance | 워라밸 만족도 (1~4) | 범주형 (Categorical) | int |
| YearsSinceLastPromotion | 마지막 승진 후 기간 | 수치형 (Numerical) | int |

## 야근을 하는 직원과 야근을 하지 않는 직원의 이직율 차이 비교
- 범주형 vs 범주형 데이터

```python
overtime_attrition = (
    pd.crosstab(hr_df["OverTime"], hr_df["Attrition"], normalize="index") * 100
)
print("야근 여부에 따른 이직 비율")
overtime_attrition
```

```python
overtime_rate = hr_df.groupby("OverTime")["Attrition"].apply(lambda x: (x == "Yes").mean() * 100)
overtime_rate
```

```python
ax = overtime_attrition.plot(kind="bar", stacked=True)

plt.title("야근 여부에 따른 이직 비율(%)")
plt.xlabel("야근여부")
plt.xticks(rotation=0)
plt.ylabel("비율(%)")
plt.show()


"""
    ***범주형과 범주형 데이터를 비교할 때, Bar차트가 적합
    야근을 하는 직원의 이직률이 야근을 하지않는 직원보다 약 20%더 높다
"""

None
```
