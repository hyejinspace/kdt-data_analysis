## Series
   - Pandas의 기본 객체 중 하나로, numpy의 ndarray를 기반으로 인덱싱 기능이 추가된
   1차원 배열을 나타낸다. (index x value의 형태)
   - index를 지정하지 않을 시, 기본적으로 ndarray와 같이 0-based 인덱스 생성한다
   - index를 지정할 경우 명시적으로 지정된 index를 사용한다.
   - 같은 타입의 0개 이상의 데이터를 가질 수 있다

## Series 함수
    - ndim : 차원 반환
    - len : 데이터 길이
    - size : 개수 반환
    - shape : 튜플의 형태로 shape반환
    - unique : 유일한 값만 udarray로 반환
    - count : NaN을 제외한 개수를 반환
    - mean : NaN를 제외한 평균
    - value_counts : NaN를 제외하고 각 값들의 빈도를 반환
## NaN(Not a Number)
- 숫자형 데이터가 들어와야 하는데 숫자형 데이터가 아닌 다른 값이 들어왔을 때 나타나는 값
- ex) DBMS -> null, JS -> undefuned, python -> None

```python
s1 = pd.Series([10, 2, 3, 3, np.nan, 2, 3, 5, 7, 8, np.nan, np.nan])

s1
```

```python
s1.size
```

```python
s1.shape
```

```python
s1.unique()
```

```python
# NaN 데이터가 있다!
print(s1.count())
print(len(s1))
```

```python
s1.value_counts()
```

```python
# head의 디폴트는 5
s1.head()
s1.head(3)
```

```python
# 이것도 디폴트는 5 (뒤에서부터 가져오기)
s1.tail()
s1.tail(1)
```

## 시리즈(Series)의 연산
- 시리즈는 시리즈끼리 연산이 가능하다. 단, index를 기준으로 연산된다.
- 이 특징은 DataFrame에서도 동일하게 적용된다.

```python
products = {
    "난리부르스 치킨": 30000,
    "요아정 벌집 요거트": 50000,
    "그릭 요거트": 13000,
    "커피": 6000,
    "떡볶이": 20000,
    "예나가 만든 샌드위치": 50000
}

s2 = pd.Series(products)
s2
```

```python
products2 = {
    "난리부르스 치킨": 30000,
    "요아정 벌집 요거트": 50000,
    "국밥": 13000,
    "라면": 6000,
    "두쫀쿠": 10000,
    "피자": 30000
}

s3 = pd.Series(products2)
s3
```

```python
s2 + s3
```

```python
s2 * 1.2
```

```python
# 가격이 3만원을 넘는 괘씸한 음식을 골라내기
price_mask = s2 > 30000
s2[price_mask]
```

```python
hobbies = {
    "승현": "칵테일방, 찜질방, 노래방",
    "예나": "요리, 노래방, 영화",
    "진성": "호프, 노래방, 호프집",
    "가나디": "그릭요거트, 그림그리기, 영화",
    "지혁": "음주, 흡연, 도박",
    "시우": "볼링, 클라이밍, 보석십자수"
}

s4 = pd.Series(hobbies)
s4
```

```python
# 악의 축 mask로 제거하기
# s4.str.contains("호프|도박|흡연")


#각 행을 접근해서 글자들이 포함되어 있으면 True,아니면 False
hobbie_mask = s4.str.contains("호프|도박|흡연")
s5 = s4[-hobbie_mask]
```

```python
evil_list = ["호프","도박","흡연"]
# @@@@@@@@@@@@@@@@@@@ 이중 반복문 컴프리헨션??????????????????????@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@
evil_mask = [sum(evil in hobby for evil in evil_list) == 0 for hobby in s4]

evil_member = s4[evil_mask]
print(evil_member)
```

```python
for hobby in s4:
    for evil in evil_list:
        print(evil in hobby)
    print()
```

## .groupby()

```python
salary_list = [
    "지혁", "희수", "승현", "우영", "희수", "시우", "슬기", "상아", "상아", "희수", "우영", "진휘", 
    "지혁", "준승", "예나", "진휘", "지호", "지혁", "희수", "승현", "시우", "준승", "우영", "예나"
]

salary_income = [
    200, 300, 400, 400, 350, 500, 300, 600, 600, 500, 600, 800,
    300, 300, 500, 500, 300, 400, 300, 800, 300, 200, 700, 500,
]
```

```python
s1 = pd.Series(data=salary_income, index=salary_list)
s1
```

```python
# .count(): 개수
s1.groupby(s1.index).count()
```

```python
s1.groupby(s1.index).sum()
```

```python
# mean: 평균
s1.groupby(s1.index).mean()
```

```python
s1.groupby(s1 > 300).count()
```

```python
s1.groupby(s1 > 300).mean()
```

```python
# 600만원 이상 받은 사용자의 수
s1.groupby(s1 >= 600).count()
```

```python
# apply: 형태를 바꾸는거
s1.groupby(s1 >= 600).apply(list)
```

```python
s1.groupby(s1 >= 600).apply(lambda x: list((x.index)))
```

```python
s1.groupby(s1 >= 600).apply(lambda x: list(zip(x.index, x)))
```

```python
# 메서드 이름에 to로 시작하는 애들은 형변환임, is 로 시작하는애들은 bool 타임 get은 getter 값을 가져오느ㄴ거 set은 setter값변환

s1.to_dict()
s1.to_list()
```
