```python
import os
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np

# 윈도우 맑은 고딕 설정
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  # 마이너스 기호 깨짐 방지

# 맥 애플고딕 설정
# plt.rcParams['font.family'] = 'AppleGothic'
# plt.rcParams['axes.unicode_minus'] = False  # 마이너스 기호 깨짐 방지

base_path = r"C:\kdt_0900_hhj\data_analysis\resource\dataset"
file_name = r"hr_employee_attrition.csv"

file_path = os.path.join(base_path, file_name)

hr_df = pd.read_csv(file_path)
hr_df
```

```python
# 4번
# 워라밸 만족도(WorkLifeBalance) 1~4점 점수별 이직률 구하고, 워라밸이 낮은 1점의 이직 위험도를 분석하기

wlb_df = hr_df.groupby("WorkLifeBalance")["Attrition"].apply(lambda x: (x == "Yes").mean()).reset_index()

wlb_df

sns.barplot(
    data=wlb_df,
    x="WorkLifeBalance",
    y="Attrition",
    hue="WorkLifeBalance",
    palette=["red","orange","yellow","blue"]
)
plt.title("워라벨 만족도에 따른 이직률")
plt.xlabel("워라벨 만족도 1-4")
plt.ylabel("이직률")
plt.show()
```

```python
# 3번
# 가설: 마지막 승진 후 기간(YearsSinceLastPromotion)이 길어질수록 이직률이 더 높아질까?

promo_df = hr_df.groupby("YearsSinceLastPromotion")["Attrition"].agg(
    YesRate=lambda x: (x == "Yes").mean() * 100,
    NoRate=lambda x: (x == "No").mean() * 100
)
promo_df

sns.lineplot(
    data=promo_df,
    x="YearsSinceLastPromotion",
    y="YesRate",
    label="이직 Yes"
)
sns.lineplot(
    data=promo_df,
    x="YearsSinceLastPromotion",
    y="NoRate",
    label="이직 No"
)

plt.title("마지막 승진 후 기간에 따른 이직률vs유지율")
plt.xlabel("마지막 승진 후 기간")
plt.ylabel("비율")
plt.show()
```

```python
# 2번
# 이직한 직원(Yes)과 잔류한 직원(No)의 월소득(MonthlyIncome) 평균 및 분포 차이 비교(boxplot)
```

```python
income_state = hr_df.groupby("Attrition")["MonthlyIncome"].agg(["count","mean","median","max","min","std"]).round(2)
income_state

sns.boxplot(
    data=hr_df,
    x="Attrition",
    y="MonthlyIncome",
    hue="Attrition",
    palette=(["pink","brown"])
)
plt.title("월소득에따른 이직 여부")
plt.xlabel("이직여부")
plt.ylabel("월 소득")
plt.show()
```

```python
# 1. 야근을 하는 직원과 야근을 하지 않는 직원의 이직율 차이 비교

overtime_attrition = hr_df.groupby("OverTime")["Attrition"].apply(lambda x: (x == "Yes").mean() * 100).round(2).reset_index()

overtime_attrition

sns.barplot(
    data=overtime_attrition,
    x="OverTime",
    y="Attrition",
    hue="Attrition",
    palette=(["blue","skyblue"])
)
```
