## 야근을 하는 직원과 야근을 하지 않는 직원의 이직율 차이 비교
- 범주형 vs 범주형 데이터

```python
overtime_attrition = (
    pd.crosstab(hr_df["OverTime"], hr_df["Attrition"], normalize="index") * 100
)
print("야근 여부에 따른 이직 비율")
overtime_attrition
```

```python
overtime_rate = hr_df.groupby("OverTime")["Attrition"].apply(lambda x: (x == "Yes").mean() * 100)
overtime_rate
```

```python
ax = overtime_attrition.plot(kind="bar", stacked=True)

plt.title("야근 여부에 따른 이직 비율(%)")
plt.xlabel("야근여부")
plt.xticks(rotation=0)
plt.ylabel("비율(%)")
plt.show()


"""
    ***범주형과 범주형 데이터를 비교할 때, Bar차트가 적합
    야근을 하는 직원의 이직률이 야근을 하지않는 직원보다 약 20%더 높다
"""

None
```

## 3번
- 가설: 마지막 승진 후 기간(YearSinceLastPromotion)이 길어질수록 이직률이 더 높아질까?
- 검증: 연속형 데이터 > 라인차트 사용하는것이 좋음 (+ 순서형도)

```python
hr_df.groupby("Attrition")["YearsSinceLastPromotion"].agg(["count","mean", "median", "min", "max", "std"])

promotion_rate = (
    pd.crosstab(hr_df["YearsSinceLastPromotion"], hr_df["Attrition"], normalize="index") * 100
)
print("마지막 승진 기간에 따른 이직율")
promotion_rate


#같은 코드??
# promotion_rate = (
#     hr_df.groupby("YearsSinceLastPromotion")["Attrition"]
#     .apply(lambda x: (x == "Yes").mean() * 100)
# )
# promotion_rate
```

```python
plt.figure(figsize=(8, 4))

# plt.plot(
#     promotion_rate.index,
#     promotion_rate.values
# )
sns.lineplot(
    data=promotion_rate,
    x=promotion_rate.index,
    y="Yes",
    label="이직률"
)

sns.lineplot(
    data=promotion_rate,
    x=promotion_rate.index,
    y="No",
    label="안이직률"
)
plt.xlabel("년수")
plt.ylabel("이직률 (%)")
plt.title("마지막 승진 후 기간에 따른 이직률")

plt.legend()
# plt.grid()

plt.show()
```

```python
# 쌤코드

promo_df = hr_df.groupby("YearsSinceLastPromotion")["Attrition"].agg(
    YesRate=lambda x: (x == "Yes").mean() * 100,
    NoRate=lambda x: (x == "No").mean() * 100,
).reset_index()

promo_df

# promo_df
plt.figure(figsize=(6, 4))

plt.plot(promo_df["YearsSinceLastPromotion"], promo_df["YesRate"], color="red", label="이직(Yes)")
plt.plot(promo_df["YearsSinceLastPromotion"], promo_df["NoRate"], color="green", label="유지(No)")
plt.title("마지막 승진 후 기간 별 이직률 vs 유지율(%)")
plt.xlabel("마지막 승진 후 기간(년)")
plt.ylabel("비율(%)")
plt.axvline(3, alpha=0.2, linestyle="--")
plt.axvline(6, alpha=0.2, linestyle="--")
plt.axvline(9, alpha=0.2, linestyle="--")
plt.legend()
plt.show()
```

# 결과
- 승진 직후 0년차에 약 18.9% 정도의 이직률을 갖는다.
- 승진 경력의 타이틀을 확보한 직원이 몸값을 올려서 이직하지 않을까?
- 1~2년 승진을 못하면 승진을 포기하고 이직을 하는 비율도 생각보다 높게 나타났다.
- 장기화 7년차(21%) 이후부터는 다시 이직률(0%)이 낮아지는 역전 현상이 일어나고있다. - 데이터 부족
- 승진 후 기간이 길어질수록 이직률이 높아진다는 부분기각!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!
